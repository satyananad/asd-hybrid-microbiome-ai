```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
import re
import warnings
warnings.filterwarnings('ignore')

# Bio Stats & Bioinformatics
from scipy.stats import shapiro, ranksums, gmean, mannwhitneyu
import statsmodels.stats.multitest as smm
from skbio.stats.ordination import pcoa
from skbio.stats.distance import permanova, DistanceMatrix
from sklearn.metrics import pairwise_distances

# Feature Selection
from sklearn.feature_selection import mutual_info_classif, RFE
from sklearn.linear_model import LogisticRegression, Lasso
from skrebate import ReliefF

# Deep Learning (Autoencoder)
import tensorflow as tf
from tensorflow.keras.models import Model as KerasModel
from tensorflow.keras.layers import Input, Dense, BatchNormalization, ReLU, Dropout
from tensorflow.keras.optimizers import Adam as KerasAdam
from tensorflow.keras.callbacks import EarlyStopping

# Graph Neural Network 
import torch
from torch.nn import Linear
import torch.nn.functional as F
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GCNConv, global_mean_pool

# ML Models & Evaluation
import optuna
from sklearn.model_selection import StratifiedKFold, train_test_split, cross_val_score
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.calibration import calibration_curve
from sklearn.manifold import TSNE
from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                             f1_score, roc_auc_score, confusion_matrix, roc_curve,
                             precision_recall_curve, average_precision_score, brier_score_loss)

# Interpretability & Stats
import shap
from statsmodels.stats.contingency_tables import mcnemar
"""
OTU abundance
     ↓
Genus aggregation -> CLR Normalization -> Low Prevalence Filter
     ↓
Statistical Analysis (Shannon, Wilcoxon, PERMANOVA)
     ↓
Feature Selection (Repeated CV Stability Selection)
     ↓
Feature Engineering (Abundance Ratios & Diversity)
     ↓
Deep Autoencoder Embedding (DL Phase)
     ↓
Microbial Correlation Graph Construction (Spearman > 0.5)
     ↓
Graph Neural Network Embedding (GNN Phase)
     ↓
Multimodal Feature Fusion (GNN + DL)
     ↓
Stacked Ensemble Classifier (LGBM, XGB, CatBoost, SVM)
     ↓
Evaluation (Holdout Set) & Post-Hoc Interpretable SHAP
"""

# 0. DATA LOADING, PARSING & DATASET SUMMARY

print("\nLoading data...")
otu = pd.read_csv("/kaggle/input/datasets/antaresnyc/human-gut-microbiome-with-asd/GSE113690_Autism_16S_rRNA_OTU_assignment_and_abundance.csv")

sample_cols = [c for c in otu.columns if c.startswith('A') or c.startswith('B')]
y_labels = np.array([1 if c.startswith('A') else 0 for c in sample_cols])
group_labels = np.array(['ASD' if y==1 else 'Control' for y in y_labels])

def extract_genus(tax_str):
    try:
        parts = str(tax_str).split(';')
        for p in parts:
            if '_g__' in p or p.strip().startswith('g__'):
                return p.replace('_g__', '').replace('g__', '').strip()
        return 'Unknown'
    except:
        return 'Unknown'

otu['Genus'] = otu['taxonomy'].apply(extract_genus)
genus_abundance = otu.groupby('Genus')[sample_cols].sum().T
genus_abundance = genus_abundance.drop(columns=['Unknown'], errors='ignore')

# Dataset Summary Output
print("\n--- Dataset Summary ---")
print(f"Total samples: {genus_abundance.shape[0]}")
print(f"ASD samples: {sum(y_labels==1)}")
print(f"Control samples: {sum(y_labels==0)}")
print(f"Original Genus features: {genus_abundance.shape[1]}")

# Low Prevalence Filtering / Normalization
prevalence = (genus_abundance > 0).mean(axis=0)
genus_filtered = genus_abundance.loc[:, prevalence >= 0.1]
print(f"Features after filtering: {genus_filtered.shape[1]}")

pseudo_count = 1e-6
pre_clr_data = genus_filtered + pseudo_count
clr_data = pre_clr_data.apply(lambda row: np.log(row / gmean(row)), axis=1)

shannon_idx = genus_filtered.apply(lambda row: -np.sum((row[row>0]/row.sum()) * np.log(row[row>0]/row.sum())), axis=1)
simpson_idx = genus_filtered.apply(lambda row: 1 - np.sum((row[row>0]/row.sum())**2), axis=1)


# 1. BIOLOGICAL STATISTICAL VALIDATION (FULL DATASET BEFORE ML)

print("\n--- 1. Statistical Microbiome Validation ---")

# Alpha Diversity
_, p_shan = mannwhitneyu(shannon_idx[y_labels==1], shannon_idx[y_labels==0])
_, p_simp = mannwhitneyu(simpson_idx[y_labels==1], simpson_idx[y_labels==0])
print(f"Alpha Diversity Mann-Whitney: Shannon p={p_shan:.4f} | Simpson p={p_simp:.4f}")

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
sns.boxplot(x=group_labels, y=shannon_idx, ax=ax[0], palette=['red', 'blue'])
ax[0].set_title(f'Shannon Diversity (p={p_shan:.4f})')
sns.boxplot(x=group_labels, y=simpson_idx, ax=ax[1], palette=['red', 'blue'])
ax[1].set_title(f'Simpson Diversity (p={p_simp:.4f})')
plt.show()

# Differential Taxa
diff_results, log2fc_list, pval_list = [], [], []
for col in clr_data.columns:
    asd_vals = clr_data[col][y_labels==1]
    ctrl_vals = clr_data[col][y_labels==0]
    stat, p_val = ranksums(asd_vals, ctrl_vals)
    diff_results.append(p_val)
    pval_list.append(p_val)
    log2fc_list.append(np.mean(asd_vals) - np.mean(ctrl_vals))
    
fdr_reject, fdr_pvals, _, _ = smm.multipletests(diff_results, alpha=0.05, method='fdr_bh')
print(f"Significantly differential taxa (FDR < 0.05): {sum(fdr_reject)}")

# Differential Table
biomarker_df = pd.DataFrame({
    'Genus': clr_data.columns, 'Log2FC': log2fc_list, 'P-Value': pval_list, 'FDR': fdr_pvals
}).sort_values(by='FDR')
print("\nTop 10 Biomarkers:")
print(biomarker_df.head(10).to_string(index=False))

# Volcano Plot
plt.figure(figsize=(7,5))
plt.scatter(log2fc_list, -np.log10(fdr_pvals), color='grey', alpha=0.5, label='Non-significant')
sig_filter = np.array(fdr_pvals) < 0.05
plt.scatter(np.array(log2fc_list)[sig_filter], -np.log10(np.array(fdr_pvals)[sig_filter]), 
            color='red', alpha=0.7, label='FDR < 0.05')
plt.axhline(-np.log10(0.05), color='black', linestyle='--', label='FDR=0.05')
plt.xlabel("Log2 Fold Change (ASD / Control in CLR)")
plt.ylabel("-Log10(FDR Corrected p)")
plt.title("Differential Abundance Volcano Plot")
plt.legend()
plt.show()

# Beta Diversity
bray_curtis_dist = pairwise_distances(genus_filtered, metric='braycurtis')
dist_matrix = DistanceMatrix(bray_curtis_dist, ids=genus_filtered.index.astype(str))
perm_result = permanova(dist_matrix, y_labels, permutations=999)
print(f"\nPERMANOVA (Bray-Curtis): pseudo-F = {perm_result['test statistic']:.2f}, p-value = {perm_result['p-value']}")

pcoa_res = pcoa(dist_matrix)
plt.figure(figsize=(6,5))
sns.scatterplot(x=pcoa_res.samples['PC1'], y=pcoa_res.samples['PC2'], 
                hue=group_labels, palette=['red', 'blue'], alpha=0.7)
plt.title(f"Beta Diversity PCoA (Bray-Curtis)\nPERMANOVA p={perm_result['p-value']}")
plt.xlabel(f"PC1 ({pcoa_res.proportion_explained['PC1']:.1%})")
plt.ylabel(f"PC2 ({pcoa_res.proportion_explained['PC2']:.1%})")
plt.show()

# 2. TRAIN-TEST SPLIT (80/20 Holdout)

X_train, X_test, y_train, y_test = train_test_split(
    clr_data, y_labels, test_size=0.20, stratify=y_labels, random_state=42
)
X_train_raw = genus_filtered.loc[X_train.index]
X_test_raw = genus_filtered.loc[X_test.index]

# 3. FEATURE SELECTION & STABILITY (TRAINING DATA ONLY)

print("\n--- 2. Feature Selection & Stability Validation ---")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
global_feature_counts = {feat: 0 for feat in X_train.columns}

fold_count = 1
for train_idx, val_idx in skf.split(X_train, y_train):
    X_train_clr_cv = X_train.iloc[train_idx]
    y_train_fold = y_train[train_idx]
    
    mi = mutual_info_classif(X_train_clr_cv, y_train_fold)
    for feat in X_train_clr_cv.columns[np.argsort(mi)[-20:]]: global_feature_counts[feat] += 1
        
    lasso = Lasso(alpha=0.01).fit(X_train_clr_cv, y_train_fold)
    for feat in X_train_clr_cv.columns[lasso.coef_ != 0]: global_feature_counts[feat] += 1
        
    relief = ReliefF(n_features_to_select=20, n_neighbors=10)
    relief.fit(X_train_clr_cv.values, y_train_fold)
    for feat in X_train_clr_cv.columns[np.argsort(relief.feature_importances_)[-20:]]: global_feature_counts[feat] += 1
    
    # Fold Proxy Accuracy Track
    proxy_model = RandomForestClassifier(random_state=42).fit(X_train_clr_cv, y_train_fold)
    fold_acc = accuracy_score(y_train[val_idx], proxy_model.predict(X_train.iloc[val_idx]))
    print(f"  Feature Selection CV Fold {fold_count}/5 Validation Proxy-Accuracy: {fold_acc:.3f}")
    fold_count += 1

final_biomarkers = [f for f, count in global_feature_counts.items() if count >= 3]
if len(final_biomarkers) < 3:
    final_biomarkers = sorted(global_feature_counts, key=global_feature_counts.get, reverse=True)[:15]

# Plot Feature Stability
feat_counts_df = pd.DataFrame({"Feature": list(global_feature_counts.keys()), "Count": list(global_feature_counts.values())})
feat_counts_df = feat_counts_df.sort_values("Count", ascending=False).head(20)

plt.figure(figsize=(8,6))
sns.barplot(x="Count", y="Feature", data=feat_counts_df, palette="viridis")
plt.title("Feature Stability Selection Across CV Folds")
plt.tight_layout()
plt.show()

# Engineering Features Helper (Includes XGBoost Column Name Fix)
def engineer_features(X_clr, X_raw, biomarkers, shannon, simpson):
    X_eng = X_clr[biomarkers].copy()
    X_eng['Shannon_Div'] = shannon.loc[X_clr.index]
    X_eng['Simpson_Div'] = simpson.loc[X_clr.index]
    top3 = biomarkers[:3]
    if len(top3) >= 2:
        for i, m1 in enumerate(top3):
            for m2 in top3[i+1:]:
                X_eng[f"Ratio_{m1}_vs_{m2}"] = np.log((X_raw[m1] + 1) / (X_raw[m2] + 1))
    
    # BUG FIX 1: Strip structural brackets for XGBoost compatibility 
    X_eng.columns = [re.sub(r'[\[\]<>]', '', col) for col in X_eng.columns]
    return X_eng

X_train_eng = engineer_features(X_train, X_train_raw, final_biomarkers, shannon_idx, simpson_idx)
X_test_eng = engineer_features(X_test, X_test_raw, final_biomarkers, shannon_idx, simpson_idx)

scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train_eng), columns=X_train_eng.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test_eng), columns=X_test_eng.columns)

# 4. DEEP LEARNING EMBEDDING (Autoencoder) 

print("\n--- 3. Training Deep Autoencoder ---")
def create_encoder(input_dim):
    inputs = Input(shape=(input_dim,))
    x = Dense(128)(inputs)
    x = BatchNormalization()(x)
    x = ReLU()(x)
    x = Dropout(0.2)(x)
    x = Dense(64)(x)
    x = BatchNormalization()(x)
    x = ReLU()(x)
    embedding = Dense(16, name="embedding_layer")(x)
    outputs = Dense(1, activation='sigmoid')(embedding)
    model = KerasModel(inputs=inputs, outputs=outputs)
    model.compile(optimizer=KerasAdam(learning_rate=0.001), loss='binary_crossentropy')
    return model

encoder_model = create_encoder(X_train_scaled.shape[1])
history = encoder_model.fit(X_train_scaled, y_train, validation_split=0.1, epochs=100, 
                  batch_size=32, callbacks=[EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)], verbose=0)

plt.figure(figsize=(6,4))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.legend()
plt.title("Autoencoder Training Convergence")
plt.show()

keras_extractor = KerasModel(inputs=encoder_model.input, outputs=encoder_model.get_layer("embedding_layer").output)
X_train_dl_emb = keras_extractor.predict(X_train_scaled, verbose=0)
X_test_dl_emb = keras_extractor.predict(X_test_scaled, verbose=0)

# 5. MICROBIAL GRAPH CONSTRUCTION & GNN EMBEDDING

print("\n--- 4. Building Microbial Graph & Training GNN ---")
correlations = X_train_scaled.corr(method='spearman')
threshold = 0.5
edges, weights = [], []

for i in range(len(correlations.columns)):
    for j in range(i + 1, len(correlations.columns)):
        weight = correlations.iloc[i, j]
        if abs(weight) > threshold:
            edges.append([i, j])
            edges.append([j, i]) 
            weights.extend([abs(weight), abs(weight)])

# Plot Correlation Heatmap
plt.figure(figsize=(10,8))
sns.heatmap(correlations, cmap="coolwarm", center=0, vmin=-1, vmax=1)
plt.title("Microbial Correlation Heatmap (Training Fold Built)")
plt.show()

# Network Visualization
G = nx.Graph()
for e in edges: G.add_edge(e[0], e[1])
plt.figure(figsize=(6,6))
nx.draw(G, node_size=50, edge_color="gray", alpha=0.6, node_color="darkred")
plt.title("Microbial Interaction Network (Spearman > 0.5)")
plt.show()

if len(edges) == 0:
    for i in range(len(correlations.columns)):
        for j in range(i + 1, len(correlations.columns)):
            edges.append([i, j])
            edges.append([j, i]) 
            weights.extend([0.1, 0.1])
            
edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()
edge_weight = torch.tensor(weights, dtype=torch.float)

def create_graph_dataset(X_df, y_array):
    data_list = []
    means = X_df.mean().values
    stds = X_df.std().values + 1e-8
    for idx in range(len(X_df)):
        raw_vals = X_df.iloc[idx].values
        z_scores = (raw_vals - means) / stds
        x = torch.tensor(np.vstack([raw_vals, z_scores]).T, dtype=torch.float)
        y = torch.tensor([y_array[idx]], dtype=torch.float)
        data = Data(x=x, edge_index=edge_index, edge_attr=edge_weight, y=y)
        data_list.append(data)
    return data_list

train_graphs = create_graph_dataset(X_train_scaled, y_train)
test_graphs = create_graph_dataset(X_test_scaled, y_test)
train_loader = DataLoader(train_graphs, batch_size=32, shuffle=True)
test_loader = DataLoader(test_graphs, batch_size=32, shuffle=False)

class MicrobiomeGNN(torch.nn.Module):
    def __init__(self, hidden_dim=32, embedding_dim=16):
        super(MicrobiomeGNN, self).__init__()
        self.conv1 = GCNConv(2, hidden_dim) 
        self.conv2 = GCNConv(hidden_dim, hidden_dim)
        self.fc_embed = Linear(hidden_dim, embedding_dim)
        self.classifier = Linear(embedding_dim, 1)

    def forward(self, x, edge_index, edge_attr, batch):
        x = F.relu(self.conv1(x, edge_index, edge_attr))
        x = F.relu(self.conv2(x, edge_index, edge_attr))
        x = global_mean_pool(x, batch)
        embedding = self.fc_embed(x)
        out = torch.sigmoid(self.classifier(F.relu(embedding)))
        return out, embedding

gnn_model = MicrobiomeGNN()
optimizer = torch.optim.Adam(gnn_model.parameters(), lr=0.005)
criterion = torch.nn.BCELoss()

gnn_model.train()
for epoch in range(40):
    for data in train_loader:
        optimizer.zero_grad()
        out, _ = gnn_model(data.x, data.edge_index, data.edge_attr, data.batch)
        loss = criterion(out.squeeze(), data.y.squeeze())
        loss.backward()
        optimizer.step()

gnn_model.eval()
def extract_gnn_embeddings(dataset):
    embeddings = []
    loader = DataLoader(dataset, batch_size=32, shuffle=False)
    with torch.no_grad():
        for data in loader:
            _, emb = gnn_model(data.x, data.edge_index, data.edge_attr, data.batch)
            embeddings.append(emb.numpy())
    return np.vstack(embeddings)

X_train_gnn_emb = extract_gnn_embeddings(train_graphs)
X_test_gnn_emb = extract_gnn_embeddings(test_graphs)

# 6. MULTIMODAL FUSION & HYBRID ENSEMBLE CLASSIFICATION

print("\n--- 5. Fusing Embeddings & Tuning Stacked Meta-Model ---")
X_train_fused = np.hstack([X_train_dl_emb, X_train_gnn_emb])
X_test_fused = np.hstack([X_test_dl_emb, X_test_gnn_emb])

# t-SNE of the Multi-Modal Fused Features
tsne = TSNE(n_components=2, random_state=42)
emb_tsne = tsne.fit_transform(X_train_fused)
plt.figure(figsize=(6,5))
plt.scatter(emb_tsne[:,0], emb_tsne[:,1], c=y_train, cmap='coolwarm', alpha=0.7)
plt.title("t-SNE Projection: Fusion Hybrid Embeddings\n(Red=ASD, Blue=Control)")
plt.show()

# Tune Final Models using Optuna
def tune_lgb(X, y):
    def obj(trial):
        p = {'n_estimators': trial.suggest_int('n_estimators', 50, 200),
             'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.1, log=True),
             'max_depth': trial.suggest_int('max_depth', 3, 10), 'verbose': -1}
        return cross_val_score(LGBMClassifier(**p), X, y, cv=3, scoring="roc_auc").mean()
    study = optuna.create_study(direction="maximize")
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    study.optimize(obj, n_trials=30)
    return study.best_params

def tune_xgb(X, y):
    def obj(trial):
        p = {'n_estimators': trial.suggest_int('n_estimators', 50, 200),
             'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.1, log=True),
             'max_depth': trial.suggest_int('max_depth', 3, 10), 'eval_metric': 'logloss'}
        return cross_val_score(XGBClassifier(**p), X, y, cv=3, scoring="roc_auc").mean()
    study = optuna.create_study(direction="maximize")
    study.optimize(obj, n_trials=30)
    return study.best_params

print("Hyperparameter Tuning Initiated via Optuna...")
lgb_params = tune_lgb(X_train_fused, y_train)
xgb_params = tune_xgb(X_train_fused, y_train)

print(f"\nBest XGBoost Parameters: {xgb_params}")
print(f"Best LightGBM Parameters: {lgb_params}")

estimators = [
    ('lgb', LGBMClassifier(**lgb_params, verbose=-1)),
    ('xgb', XGBClassifier(**xgb_params)),
    ('cat', CatBoostClassifier(iterations=100, verbose=0)),
    ('svm', SVC(probability=True))
]

hybrid_stack = StackingClassifier(estimators=estimators, final_estimator=LogisticRegression())
hybrid_stack.fit(X_train_fused, y_train)

rf_base = RandomForestClassifier(random_state=42).fit(X_train_scaled, y_train)
svm_base = SVC(probability=True, random_state=42).fit(X_train_scaled, y_train)
xgb_base = XGBClassifier(eval_metric='logloss').fit(X_train_scaled, y_train)

dl_stack = StackingClassifier(estimators=estimators, final_estimator=LogisticRegression()).fit(X_train_dl_emb, y_train)
gnn_stack = StackingClassifier(estimators=estimators, final_estimator=LogisticRegression()).fit(X_train_gnn_emb, y_train)

# 7. FINAL EVALUATION REPORT (ABLATION TABLE) & STATS

print("\n--- 6. Test Set Hybrid Multimodal Ablation Study (20% Holdout) ---")
final_results = {}
def evaluate_model(name, model, X_eval, y_true):
    prob = model.predict_proba(X_eval)[:, 1]
    pred = model.predict(X_eval)
    final_results[name] = {
        'ACC': accuracy_score(y_true, pred), 'PREC': precision_score(y_true, pred), 'REC': recall_score(y_true, pred),
        'F1': f1_score(y_true, pred), 'AUC': roc_auc_score(y_true, prob), 'Preds': pred, 'Probs': prob
    }

evaluate_model('Random Forest', rf_base, X_test_scaled, y_test)
evaluate_model('SVM', svm_base, X_test_scaled, y_test)
evaluate_model('XGBoost', xgb_base, X_test_scaled, y_test)
evaluate_model('DL (Autoencoder+ML)', dl_stack, X_test_dl_emb, y_test)
evaluate_model('GNN (Graph+ML)', gnn_stack, X_test_gnn_emb, y_test)
evaluate_model('SGHD (Hybrid GNN+DL+ML)', hybrid_stack, X_test_fused, y_test)

print(f"{'Model Name':<25} | {'Type':<10} | {'ROC-AUC':<8} | {'Accuracy':<8} | {'F1':<8}")
print("-" * 72)
for m, m_type in zip(final_results.keys(), ['ML','ML','ML','DL','Graph','Hybrid']):
    res = final_results[m]
    print(f"{m:<25} | {m_type:<10} | {res['AUC']:.4f}   | {res['ACC']:.4f}   | {res['F1']:.4f}")

# Model Comparison Bar Chart
models = list(final_results.keys())
auc_scores = [final_results[m]['AUC'] for m in models]
plt.figure(figsize=(10,5))
sns.barplot(x=auc_scores, y=models, palette="viridis")
plt.title("Benchmarking Models by ROC-AUC (Test Holdout)")
plt.xlim((0.5, 1.0))
plt.show()

print("\n--- 7. Medical Statistical Validation ---")
# McNemar Test Formulation BugFix implemented
def calculate_mcnemar(y_true, pred1, pred2):
    table = [[0,0],[0,0]]
    for t, p1, p2 in zip(y_true, pred1, pred2):
        if p1 == t and p2 == t: table[0][0]+=1
        elif p1 == t and p2 != t: table[0][1]+=1
        elif p1 != t and p2 == t: table[1][0]+=1
        else: table[1][1]+=1
    return mcnemar(table, exact=False, correction=True).pvalue

p_rf = calculate_mcnemar(y_test, final_results['SGHD (Hybrid GNN+DL+ML)']['Preds'], final_results['Random Forest']['Preds'])
print(f"McNemar Test -> Hybrid vs RF Baseline p-value: {float(np.squeeze(p_rf)):.4f}")

# Bootstrap AUC CI 
def bootstrap_auc(y_true, y_prob, n_bootstraps=1000):
    np.random.seed(42)
    scores = []
    y_true_arr = np.array(y_true)
    y_prob_arr = np.array(y_prob)
    for _ in range(n_bootstraps):
        idx = np.random.randint(0, len(y_prob), len(y_prob))
        if len(np.unique(y_true_arr[idx])) < 2: continue
        scores.append(roc_auc_score(y_true_arr[idx], y_prob_arr[idx]))
    scores.sort()
    return scores[int(0.025 * len(scores))], scores[int(0.975 * len(scores))]

auc_l, auc_u = bootstrap_auc(y_test, final_results['SGHD (Hybrid GNN+DL+ML)']['Probs'])
sghd_prob = final_results['SGHD (Hybrid GNN+DL+ML)']['Probs']
sghd_pred = final_results['SGHD (Hybrid GNN+DL+ML)']['Preds']

print(f"Final Hybrid Test ROC-AUC: {final_results['SGHD (Hybrid GNN+DL+ML)']['AUC']:.3f} (95% CI: {auc_l:.3f}–{auc_u:.3f})")

# 8. POST-EVALUATION VISUALIZATIONS (Confusion, ROC Panels, SHAP)

# Final Confusion Matrix Plot 
cm = confusion_matrix(y_test, sghd_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', xticklabels=['Control','ASD'], yticklabels=['Control','ASD'])
plt.title("Test Set Confusion Matrix (Proposed Hybrid)")
plt.ylabel('True Label'), plt.xlabel('Predicted Label')
plt.show()

# Multi-Curve Plot Grid (ROC, PR, Calibration)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fpr, tpr, _ = roc_curve(y_test, sghd_prob)
axes[0].plot(fpr, tpr, label=f"Hybrid AUC = {final_results['SGHD (Hybrid GNN+DL+ML)']['AUC']:.3f}", color='darkred', lw=2)
axes[0].plot([0, 1], [0, 1], 'k--')
axes[0].set_title('Test Set ROC Curve')

pr, rec, _ = precision_recall_curve(y_test, sghd_prob)
axes[1].plot(rec, pr, label=f"AP = {average_precision_score(y_test, sghd_prob):.3f}", color='navy', lw=2)
axes[1].set_title('Test Set Precision-Recall Curve')

prob_true, prob_pred = calibration_curve(y_test, sghd_prob, n_bins=10)
axes[2].plot(prob_pred, prob_true, marker='o', color='forestgreen', label=f"Brier = {brier_score_loss(y_test, sghd_prob):.3f}")
axes[2].plot([0, 1], [0, 1], 'k--')
axes[2].set_title('Test Set Calibration Curve')
for ax in axes: ax.legend()
plt.tight_layout()
plt.show()

# Complete SHAP Explainability Model
print("\n--- 8. Generating Explanability Interpretations (SHAP) ---")
# The structural XGBoost ValueError is resolved upstream in `engineer_features` through regex striping, mapping exactly onto the SHAP matrix here
surrogate = XGBClassifier().fit(X_train_scaled, y_train)
explainer = shap.TreeExplainer(surrogate)
plt.figure(figsize=(10, 8))
shap.summary_plot(explainer.shap_values(X_test_scaled), X_test_scaled, show=False)
plt.title("SHAP Biomarker Concept Explainer (Test Data Surrogate)")
plt.tight_layout()
plt.show()
```
